```python
l1 = [34,54,32,33,34,32,59,67]
Even_list = [] # empty list to store even item
Even_counter = 0 # empty variable to store count of even items
Odd_list = [] #empty list to store odd items
Odd_counter = 0 # emptty variable to store count of odd items
for i in l1: # to itrate over a list we apply for loop 
    if i%2==0: # even cond T
        Even_list.append(i) # add item into even list
        Even_counter+=1 # increase even counter 

    else:
        Odd_list.append(i) # add item into odd list
        Odd_counter+=1# increse odd counter

print(l1)
print(f"Total numbers of even items {Even_counter} \nlist of these numbers {Even_list}")
print(f"Total numbers of Odd items {Odd_counter} \nlist of these numbers {Odd_list}")
    
```
